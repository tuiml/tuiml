# End-to-End Machine Learning Workflow: Diabetes Prediction

This tutorial demonstrates a complete, professional machine learning workflow using the **TuiML** library. We will build a model to predict diabetes using the classic Pima Indians Diabetes Database.

The workflow follows these structured steps:
1. **Problem Identification**
2. **Data Collection and Exploration (EDA)**
3. **Data Preprocessing**
4. **Feature Engineering and Selection**
5. **Dataset Splitting**
6. **Model Training and Selection using Benchmarks**
7. **Comparative Analysis**
8. **Model Deployment (Serving)**
9. **Calling the Model API**
10. **The Same Pipeline, Two Lines**: `Workflow` and `train()`

## 1. Problem Identification

**Objective:** Predict whether a patient is diabetic based on medical diagnostic measurements.

**Task Type:** Binary Classification (Diabetic: 1, Non-diabetic: 0).

**Dataset Features:**
- **Pregnancies**: Number of times pregnant
- **Glucose**: Plasma glucose concentration
- **BloodPressure**: Diastolic blood pressure (mm Hg)
- **SkinThickness**: Triceps skin fold thickness (mm)
- **Insulin**: 2-Hour serum insulin (mu U/ml)
- **BMI**: Body mass index (weight in kg/(height in m)^2)
- **DiabetesPedigreeFunction**: Diabetes pedigree function
- **Age**: Age (years)
- **Outcome**: Class variable (target)

## 2. Data Collection and Exploration

We use TuiML's built-in dataset loaders to pull the diabetes data.

````python
from tuiml.datasets import load_diabetes
import pandas as pd

# Load dataset
dataset = load_diabetes()
X, y = dataset.X, dataset.y
feature_names = dataset.feature_names

# Convert to Pandas for easy EDA
df = dataset.to_pandas()

print(f"Dataset Shape: {df.shape}")
df.head()
````

### Exploratory Data Analysis (EDA)
Let's look at the statistical summary and class distribution.

````python
print("Feature Statistics:")
display(df.describe().T)

print("\nClass Distribution (0 = Healthy, 1 = Diabetic):")
print(df['target'].value_counts(normalize=True))
````

## 3. Data Preprocessing

In many medical datasets, 0 values often represent missing data (e.g., biological impossibility of 0 Glucose or 0 BMI). We will:
1. Replace invalid 0s with NaNs.
2. Impute missing values using the mean.
3. Standardize features for gradient-sensitive algorithms.

````python
import numpy as np
from tuiml.preprocessing.imputation import SimpleImputer
from tuiml.preprocessing.scaling import StandardScaler

# 1. Handle "missing" values (0s in Glucose, BP, BMI, etc.)
X_prep = X.copy()
for i in [1, 2, 5]:  # Indices for plas, pres, mass
    X_prep[X_prep[:, i] == 0, i] = np.nan

# 2. Impute NaNs with Mean
imputer = SimpleImputer(strategy="mean")
X_imputed = imputer.fit_transform(X_prep)

# 3. Standardize (Z-Score)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_imputed)

print(f"Preprocessed shape: {X_scaled.shape}")
````

## 4. Feature Engineering and Selection

We'll use the **ANOVA F-test (`f_classif`)** to select the top 5 most predictive features. This is a robust method for continuous numerical features in classification tasks.

````python
from tuiml.features.selection import SelectKBestSelector
from tuiml.evaluation.metrics import f_classif

# Select top 5 features using ANOVA F-test
selector = SelectKBestSelector(score_func=f_classif, k=5)
X_selected = selector.fit_transform(X_scaled, y)

selected_indices = selector.get_support(indices=True)
selected_features = [feature_names[i] for i in selected_indices]

print("Top 5 Selected Features:")
for feat in selected_features:
    print(f" - {feat}")

print(f"\nSelected feature matrix shape: {X_selected.shape}")
````

## 5. Class Balancing with SMOTE

Now we apply SMOTE to balance the classes on the selected features.

````python
from tuiml.preprocessing.sampling import SMOTESampler

# Apply SMOTE on the selected features
smote = SMOTESampler(random_state=42, k_neighbors=2)
X_resampled, y_resampled = smote.fit_resample(X_selected, y)

print(f"Original shape: {X_selected.shape}")
print(f"Resampled shape: {X_resampled.shape}")
print(f"\nClass distribution after SMOTE:")
unique, counts = np.unique(y_resampled, return_counts=True)
for u, c in zip(unique, counts):
    print(f"  Class {int(u)}: {c} samples")
````

## 6. Dataset Splitting

We use a stratified holdout split for final validation.

````python
from tuiml.evaluation.splitting import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_resampled, y_resampled, 
    test_size=0.2, 
    stratify=y_resampled, 
    random_state=42
)

print(f"Training set: {X_train.shape}")
print(f"Test set: {X_test.shape}")
````

## 7. Model Training and Selection using Benchmarks

TuiML's `Benchmark` class runs multiple models across datasets and validation schemes systematically, and returns a `BenchmarkResult` with tables, rankings, and statistical tests.

````python
import tuiml
from tuiml.algorithms.trees import RandomForestClassifier
from tuiml.algorithms.gradient_boosting import XGBoostClassifier
from tuiml.algorithms.svm import SVC

# Candidate models as strict {"name", "params"} specs; "label" sets the display name
model_specs = [
    {"name": "RandomForestClassifier", "params": {"n_estimators": 100}, "label": "Random Forest"},
    {"name": "XGBoostClassifier", "params": {"n_estimators": 100}, "label": "XGBoost"},
    {"name": "SVC", "label": "SVM (SVC)"},
]

# The same models as configured instances, for the final retrain later
models = {
    "Random Forest": RandomForestClassifier(n_estimators=100),
    "XGBoost": XGBoostClassifier(n_estimators=100),
    "SVM (SVC)": SVC(),
}

# 5-fold stratified cross-validation on the training data
# Metrics are the exact function names from tuiml.evaluation.metrics
result = tuiml.Benchmark(
    models=model_specs,
    datasets=[{"name": "Diabetes_Train", "X": X_train, "y": y_train}],
    evaluation={
        "cv": 5,
        "stratify": True,
        "metrics": ["accuracy_score", "f1_score", "roc_auc_score"],
    },
    random_seed=42,
).run()

print("\nBenchmark completed.")
````

## 8. Comparative Analysis

We can now view a high-level summary and detailed metrics table for all models.

````python
print("=== Benchmark Summary ===")
print(result.summary())

print("\n=== Performance Matrix (Markdown) ===")
print(result.to_markdown())

# Find the best model based on accuracy
best = result.best(metric="accuracy_score")
best_model_name = best.loc["Diabetes_Train", "best_model"]
print(f"\nBest model: {best_model_name}")
````

## 9. Model Deployment (Serving)

Finally, we train the best model on the full training set and serve it.

`save` and `serve` are **methods on the model**: there are no `tuiml.save()` / `tuiml.load()` top-level functions.

````python
import tuiml

# Retrain final model on full resampled data
final_model = models[best_model_name]
final_model.fit(X_resampled, y_resampled)

# Save the model and its metadata - save() is a method on the model
final_model.save("diabetes_model.pkl", metadata={
    "workflow": "diabetes_prediction_v1",
    "selected_features": selected_features,
    "feature_indices": selected_indices.tolist(),
})

print(f"Model '{best_model_name}' saved as 'diabetes_model.pkl'")
print(f"Selected features: {selected_features}")
````

````python
# Start the REST API server (runs in background in Jupyter)
final_model.serve(port=8001)

# Or run from your terminal:
# tuiml serve diabetes_model.pkl --port 8001
````

## 10. Calling the Model API

Once the server is running, you can make predictions via HTTP requests. Below are examples using both **curl** (command line) and **Python requests**.

### Available Endpoints:
| Endpoint | Method | Description |
|----------|--------|-------------|
| `/models` | GET | List loaded models |
| `/models/{model_id}` | GET | Get model info |
| `/models/{model_id}/predict` | POST | Make predictions |
| `/models/{model_id}/predict_proba` | POST | Get class probabilities |
| `/health` | GET | Server health check |
| `/docs` | GET | Interactive API documentation (Swagger UI) |

### Using curl (Command Line)

```bash
# Check loaded models
curl http://127.0.0.1:8001/models

# Make a prediction (5 features: preg, plas, mass, pedi, age)
curl -X POST http://127.0.0.1:8001/models/RandomForestClassifier/predict \
  -H "Content-Type: application/json" \
  -d '{"features": [[6.0, 148.0, 33.6, 0.627, 50.0]]}'

# Get prediction probabilities
curl -X POST http://127.0.0.1:8001/models/RandomForestClassifier/predict_proba \
  -H "Content-Type: application/json" \
  -d '{"features": [[6.0, 148.0, 33.6, 0.627, 50.0], [1.0, 85.0, 26.6, 0.351, 31.0]]}'
```

### Using Python requests

````python
import requests

BASE_URL = "http://127.0.0.1:8001"
MODEL_ID = "RandomForestClassifier"

# Sample data (using selected features: preg, plas, mass, pedi, age)
sample_data = {
    "features": [
        [6.0, 148.0, 33.6, 0.627, 50.0],  # High-risk patient
        [1.0, 85.0, 26.6, 0.351, 31.0],   # Lower-risk patient
        [0.0, 100.0, 22.0, 0.200, 25.0],  # Young, healthy BMI
    ]
}

# Make predictions
response = requests.post(
    f"{BASE_URL}/models/{MODEL_ID}/predict",
    json=sample_data
)
print("=== Predictions ===")
print(response.json())
````

````python
# Get prediction probabilities
response = requests.post(
    f"{BASE_URL}/models/{MODEL_ID}/predict_proba",
    json=sample_data
)
result = response.json()

print("=== Prediction Probabilities ===")
print(f"Classes: {result['classes']} (0=Healthy, 1=Diabetic)\n")

for i, probs in enumerate(result['probabilities']):
    print(f"Patient {i+1}: P(Healthy)={probs[0]:.1%}, P(Diabetic)={probs[1]:.1%}")
````

## 11. The Same Pipeline, Two Lines

The manual steps above are worth doing once, to see what happens at each stage. In practice you write the same thing as a pipeline and let TuiML handle the train/test bookkeeping.

Both options below produce a **fitted `Workflow`**: a pipeline that *is* a model, so `predict` / `score` / `save` / `serve` are methods on it, and every transformation travels with it.

### Option A: `Workflow`, the pipeline as a list of steps

A `Workflow` is an ordered list of steps whose **last step is the model**, and every step is a **configured instance** of an imported class. This is the object door into TuiML: imports, autocomplete, type checking. There is no chaining and no hidden state; what you wrote down is the pipeline.

`fit()` is the only execution method. It always leaves the pipeline fitted on all the data, and passing `cv=` additionally measures held-out performance first.

````python
from tuiml.workflow import Workflow
from tuiml.preprocessing import StandardScaler, SMOTESampler
from tuiml.features.selection import SelectKBestSelector
from tuiml.algorithms.trees import RandomForestClassifier

# The same four stages as sections 3-5, written as one list of steps
wf = Workflow([
    StandardScaler(),                          # 1. z-score
    SelectKBestSelector(k=5),                  # 2. top 5 features
    SMOTESampler(),                            # 3. balance classes
    RandomForestClassifier(n_estimators=100),  # 4. model
])

# One call: 5-fold CV for the estimate, then refit on every row
wf.fit(X_imputed, y, cv=5, random_seed=42)

print(f"Steps: {list(wf.named_steps)}")
print(f"Workflow Score (Accuracy): "
      f"{wf.metrics_['cv_accuracy_score_mean']:.4f} ± {wf.metrics_['cv_accuracy_score_std']:.4f}")

# Serve the whole pipeline - scaling and selection included - straight from the object
wf.serve(port=8002)
````

### Option B: `train()`, the same pipeline as one declarative spec

`train()` is the declarative door: it takes **exactly one argument**, a spec dict (or a path to a `.json` file holding it). Every component in the spec, the model and each pipeline step, is a `{"name": ..., "params": {...}}` dict, which makes the whole run serializable to JSON: the form an LLM agent emits. It returns the identical fitted `Workflow`.

Note the shape: hyperparameters live under `"params"`, all feature engineering lives in one ordered `"pipeline"` list, and evaluation options are grouped under `"evaluation"`.

````python
import tuiml

# The complete pipeline, feature selection and SMOTE included, as one spec dict
model = tuiml.train({
    "model": {"name": "RandomForestClassifier", "params": {"n_estimators": 100}},
    "data": {"X": X_imputed, "y": y},
    "pipeline": [
        {"name": "StandardScaler"},
        {"name": "SelectKBestSelector", "params": {"k": 5}},
        {"name": "SMOTESampler"},
    ],
    "evaluation": {"cv": 5},
    "random_seed": 42,
})

print(f"API Score (Accuracy): "
      f"{model.metrics_['cv_accuracy_score_mean']:.4f} ± {model.metrics_['cv_accuracy_score_std']:.4f}")

# The fitted final estimator, if you want it directly
print(f"Final estimator: {type(model.model_).__name__}")

# to_config() round-trips it back to the spec that would reproduce this run
print(f"Spec: {model.to_config()}")

# Serve it
tuiml.serve(model, port=8003)
````

### API shapes worth remembering

| What you want | How to write it |
|---|---|
| A pipeline | `Workflow([step, step, ..., model])`: steps are configured instances, the last one is the model |
| Fit it | `wf.fit(data, target=..., cv=10)`: the only execution method |
| Its results | `wf.metrics_`, `wf.cv_results_`, `wf.model_`, `wf.metadata_` |
| Use it | `wf.predict(X)`, `wf.score(X, y)`, `wf.evaluate(X, y)` |
| Persist it | `wf.save(path)` / `Workflow.load(path)` |
| The same run as data | `tuiml.train({"model": {...}, "data": {...}, "pipeline": [...], "evaluation": {...}})` |

**Note on Hyperparameters**: We used `n_estimators=100` for both Random Forest and XGBoost. TuiML follows standard naming conventions for consistency across different algorithms.
